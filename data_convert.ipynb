```python
from pathlib import Path
import pandas as pd
import pyreadstat
```

```python

sav_path = Path("LAL 2025.sav")

# 1. Raw version: keeps original numeric/coded values
df_raw, meta = pyreadstat.read_sav(
    sav_path,
    apply_value_formats=False
)

# 2. Labelled version: replaces coded values with human-readable labels
df_labelled, meta_labelled = pyreadstat.read_sav(
    sav_path,
    apply_value_formats=True
)

print("Raw shape:", df_raw.shape)
print("Labelled shape:", df_labelled.shape)

df_raw.head()
```

```python
variable_view = pd.DataFrame({
    "variable_name": meta.column_names,
    "variable_label": meta.column_labels,
    "type": [meta.readstat_variable_types.get(col) for col in meta.column_names],
    "format": [meta.original_variable_types.get(col) for col in meta.column_names],
    "measure": [meta.variable_measure.get(col) for col in meta.column_names],
    "missing_ranges": [meta.missing_ranges.get(col) for col in meta.column_names],
    "missing_user_values": [meta.missing_user_values.get(col) for col in meta.column_names],
    "value_label_set": [meta.variable_to_label.get(col) for col in meta.column_names]
})
variable_view
```

```python
def get_value_labels_for_variable(variable_name):
    label_set_name = meta.variable_to_label.get(variable_name)

    if label_set_name is None:
        return ""

    labels_dict = meta.value_labels.get(label_set_name, {})

    return "; ".join(
        f"{code} = {label}"
        for code, label in labels_dict.items()
    )

variable_view = pd.DataFrame({
    "variable_name": meta.column_names,
    "variable_label": meta.column_labels,
    "value_label_set": [
        meta.variable_to_label.get(col, "")
        for col in meta.column_names
    ],
    "value_labels": [
        get_value_labels_for_variable(col)
        for col in meta.column_names
    ]
})


# Save as CSV
variable_view.to_csv(
    "variable_view_with_value_labels.csv",
    index=False,
    encoding="utf-8-sig"
)

# Save as Excel
variable_view.to_excel(
    "variable_view_with_value_labels.xlsx",
    index=False
)

variable_view
```
